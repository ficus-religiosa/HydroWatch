```python
import os, json
from pathlib import Path
from PIL import Image
import random

def find_coco_jsons(root):
    """Walk the tree, return paths to files that look like COCO-format annotation files."""
    found = []
    for dirpath, _, filenames in os.walk(root):
        for f in filenames:
            if f.endswith(".json"):
                full = os.path.join(dirpath, f)
                try:
                    with open(full) as fh:
                        data = json.load(fh)
                    if "images" in data and "annotations" in data:
                        found.append(full)
                except (json.JSONDecodeError, UnicodeDecodeError):
                    print(f"  [skip — not valid/readable JSON] {full}")
    return found

def inspect_coco(json_path, root, sample_n=50):
    with open(json_path) as f:
        data = json.load(f)

    n_images = len(data["images"])
    n_anns = len(data["annotations"])
    cats = [c["name"] for c in data.get("categories", [])]

    # declared sizes from the json itself
    declared_sizes = {(img.get("width"), img.get("height")) for img in data["images"][:sample_n]}

    # bbox vs segmentation check
    sample_anns = data["annotations"][:sample_n]
    has_bbox = sum(1 for a in sample_anns if a.get("bbox"))
    has_seg = sum(1 for a in sample_anns if a.get("segmentation"))

    print(f"\n--- {json_path} ---")
    print(f"Images (json entries): {n_images}")
    print(f"Annotations: {n_anns}")
    print(f"Classes ({len(cats)}): {cats}")
    print(f"Declared sizes (sample of {sample_n}): {declared_sizes}")
    print(f"Sample of {sample_n} anns — with bbox: {has_bbox}, with segmentation: {has_seg}")

    # try to actually open a few real image files to verify sizes on disk match the json
    checked = 0
    mismatches = 0
    for img in random.sample(data["images"], min(10, len(data["images"]))):
        fname = img["file_name"]
        candidates = list(Path(root).rglob(os.path.basename(fname)))
        if candidates:
            try:
                real_size = Image.open(candidates[0]).size  # (width, height)
                json_size = (img.get("width"), img.get("height"))
                checked += 1
                if real_size != json_size:
                    mismatches += 1
                    print(f"  SIZE MISMATCH: {fname} — json says {json_size}, actual file is {real_size}")
            except Exception as e:
                print(f"  Could not open {fname}: {e}")
        else:
            print(f"  FILE NOT FOUND on disk: {fname}")
    print(f"Checked {checked}/10 sampled files against disk, {mismatches} mismatches")

def count_raw_images(root, exts=(".jpg", ".jpeg", ".png")):
    total = 0
    for dirpath, _, filenames in os.walk(root):
        total += sum(1 for f in filenames if f.lower().endswith(exts))
    return total


for label, root in [
    ("DS1 - TrashCan", r"D:\zaynu\Documents\coding\PS\3-1\HydroWatch\datasets\DS1"),
    ("DS3 - SeaClear", r"D:\zaynu\Documents\coding\PS\3-1\HydroWatch\datasets\DS3"),
]:
    print(f"\n{'='*60}\n{label}\n{'='*60}")
    raw_count = count_raw_images(root)
    print(f"Raw image files found on disk (all subfolders): {raw_count}")

    jsons = find_coco_jsons(root)
    print(f"COCO-format JSON files found: {jsons}")
    for j in jsons:
        inspect_coco(j, root)
```

```python
import json, os
from collections import defaultdict, Counter

root = r"D:\zaynu\Documents\coding\PS\3-1\HydroWatch\datasets\DS3"

with open(f"{root}\\dataset.json") as f:
    data = json.load(f)

cat_id_to_name = {c["id"]: c["name"] for c in data["categories"]}
img_id_to_file = {img["id"]: img["file_name"] for img in data["images"]}

# --- Q1: what metadata fields exist per image? (may reveal dive/timestamp/camera fields) ---
print("=== Q1: image record fields ===")
print(json.dumps(data["images"][0], indent=2))
print("\nSample file_name values:")
for img in data["images"][:5]:
    print(" ", img["file_name"])

# --- Q2: image + annotation counts per site-camera folder ---
print("\n=== Q2: per site-camera counts ===")
folder_imgs = Counter()
folder_anns = Counter()
imgid_to_folder = {}

for img in data["images"]:
    parts = img["file_name"].replace("\\", "/").split("/")
    folder = "/".join(parts[:-1]) if len(parts) > 1 else "(root)"
    folder_imgs[folder] += 1
    imgid_to_folder[img["id"]] = folder

for ann in data["annotations"]:
    folder_anns[imgid_to_folder.get(ann["image_id"], "?")] += 1

for folder in sorted(folder_imgs):
    print(f"  {folder}: {folder_imgs[folder]} imgs, {folder_anns[folder]} anns")

# --- Q3: do basenames repeat across cameras within the same site? (simultaneous capture test) ---
print("\n=== Q3: basename collisions across cameras within a site ===")
site_to_cam_basenames = defaultdict(lambda: defaultdict(set))
for img in data["images"]:
    parts = img["file_name"].replace("\\", "/").split("/")
    if len(parts) >= 3:
        site, cam, fname = parts[0], parts[1], parts[-1]
        site_to_cam_basenames[site][cam].add(os.path.splitext(fname)[0])

for site, cams in site_to_cam_basenames.items():
    camlist = list(cams)
    print(f"\n  {site} — cameras: {camlist}")
    for i in range(len(camlist)):
        for j in range(i+1, len(camlist)):
            shared = cams[camlist[i]] & cams[camlist[j]]
            print(f"    {camlist[i]} ∩ {camlist[j]}: {len(shared)} shared basenames")
            if shared:
                print(f"      examples: {list(shared)[:5]}")

# --- Q4: class distribution per SITE (does a held-out site even contain the classes?) ---
print("\n=== Q4: class coverage per site ===")
site_classes = defaultdict(Counter)
for ann in data["annotations"]:
    folder = imgid_to_folder.get(ann["image_id"], "?")
    site = folder.split("/")[0]
    site_classes[site][cat_id_to_name[ann["category_id"]]] += 1

for site in sorted(site_classes):
    c = site_classes[site]
    print(f"\n  {site}: {len(c)}/40 classes present, {sum(c.values())} total anns")
    print(f"    top 5: {c.most_common(5)}")
    missing = set(cat_id_to_name.values()) - set(c)
    print(f"    missing {len(missing)} classes")

# --- Q5: filename sequencing within one camera folder (frame-adjacency risk) ---
print("\n=== Q5: sorted filenames in one camera folder ===")
first_folder = sorted(folder_imgs)[0]
names = sorted(f for f in (img["file_name"] for img in data["images"]) if f.startswith(first_folder))
print(f"  {first_folder} — first 15:")
for n in names[:15]:
    print("   ", os.path.basename(n))
```

```python
import json, os
from collections import defaultdict, Counter
from PIL import Image

root = r"D:\zaynu\Documents\coding\PS\3-1\HydroWatch\datasets\DS3"

loc = {}
for dirpath, _, files in os.walk(root):
    parts = os.path.relpath(dirpath, root).split(os.sep)
    for f in files:
        if f.lower().endswith((".jpg", ".jpeg", ".png")):
            if len(parts) >= 2:
                loc[f] = (parts[0], parts[1])

with open(f"{root}\\dataset.json") as f:
    data = json.load(f)
cat = {c["id"]: c["name"] for c in data["categories"]}
imgid_loc = {i["id"]: loc.get(i["file_name"], ("?", "?")) for i in data["images"]}
imgid_name = {i["id"]: i["file_name"] for i in data["images"]}

folder_imgs = defaultdict(list)
for i in data["images"]:
    folder_imgs[imgid_loc[i["id"]]].append(i)

folder_cls = defaultdict(Counter)
folder_ann_area = defaultdict(list)
for a in data["annotations"]:
    k = imgid_loc[a["image_id"]]
    folder_cls[k][cat[a["category_id"]]] += 1
    if a.get("bbox"):
        folder_ann_area[k].append(a["bbox"][2] * a["bbox"][3])

DEBRIS_EXCL = {c for c in cat.values()
               if c.startswith(("animal_", "rov_")) or c in {"plant", "unknown_instance"}}

for k in sorted(folder_imgs):
    imgs = folder_imgs[k]
    cls = folder_cls[k]
    anns = sum(cls.values())
    debris = sum(v for c, v in cls.items() if c not in DEBRIS_EXCL)

    # declared resolutions
    res = Counter((i["width"], i["height"]) for i in imgs)

    # verify a few against disk
    real_res = Counter()
    for i in imgs[:5]:
        p = os.path.join(root, k[0], k[1], i["file_name"])
        if os.path.exists(p):
            real_res[Image.open(p).size] += 1

    areas = folder_ann_area[k]
    med_area = sorted(areas)[len(areas)//2] if areas else 0
    img_area = imgs[0]["width"] * imgs[0]["height"]

    print(f"\n{'='*70}")
    print(f"{k[0]} / {k[1]}")
    print(f"{'='*70}")
    print(f"  images: {len(imgs)}   annotations: {anns}   ann/img: {anns/len(imgs):.2f}")
    print(f"  debris: {debris}  |  non-debris: {anns - debris}  ({100*debris/anns:.1f}% debris)")
    print(f"  classes present: {len(cls)}/40")
    print(f"  declared resolution: {dict(res)}")
    print(f"  verified on disk (5 sampled): {dict(real_res)}")
    print(f"  median bbox area: {med_area:.0f} px ({100*med_area/img_area:.3f}% of frame)")
    print(f"  class counts:")
    for c, n in cls.most_common():
        print(f"      {n:>6}  {c}")
```

```python
import json, re
from collections import Counter

base = r"D:\zaynu\Documents\coding\PS\3-1\HydroWatch\datasets\DS1\instance_version"
train = [i["file_name"] for i in json.load(open(f"{base}\\instances_train_trashcan.json"))["images"]]
val   = [i["file_name"] for i in json.load(open(f"{base}\\instances_val_trashcan.json"))["images"]]

print("Sample train:", train[:8])
print("Sample val:  ", val[:8])
print("Exact filename overlap:", len(set(train) & set(val)))

def clip(f):
    m = re.match(r"(.+?)[_-]?\d+\.(jpg|png|jpeg)$", f, re.I)
    return m.group(1) if m else f

tc, vc = {clip(f) for f in train}, {clip(f) for f in val}
print(f"Train clips: {len(tc)}, Val clips: {len(vc)}, shared: {len(tc & vc)}")
```
